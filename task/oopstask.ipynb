##  Encapsulation
### Task: Employee Salary
Create an `Employee` class with private `__salary`, `get_salary()`, and `set_salary()` with validation.

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name          # public
        self.__salary = salary    # private

    def get_salary(self):
        return self.__salary

    def set_salary(self, new_salary):
        if new_salary < 0:
            print("Invalid salary")
        else:
            self.__salary = new_salary


employee = Employee("Priya", 30000)

employee.set_salary(35000)
print(employee.get_salary())   # Expected: 35000

employee.set_salary(-5000)     # Expected: Invalid salary
```

##  Inheritance
### Task 1 — Single Inheritance: Person → Student

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display_person(self):
        print(f"Name: {self.name}")
        print(f"Age : {self.age}")


class Student(Person):
    def __init__(self, name, age, student_id, course):
        super().__init__(name, age)
        self.student_id = student_id
        self.course = course

    def display_student(self):
        self.display_person()
        print(f"Student ID: {self.student_id}")
        print(f"Course    : {self.course}")


s1 = Student("Arun", 20, "S101", "Python OOP")
s1.display_student()
```

### Task 2 — Hierarchical Inheritance: Person → Doctor / Patient

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display_info(self):
        print(f"Name: {self.name} | Age: {self.age}")


class Doctor(Person):
    def __init__(self, name, age, specialization):
        super().__init__(name, age)
        self.specialization = specialization

    def diagnose(self):
        self.display_info()
        print(f"Specialization: {self.specialization}")
        print(f"Dr. {self.name} is diagnosing the patient.")


class Patient(Person):
    def __init__(self, name, age, disease):
        super().__init__(name, age)
        self.disease = disease

    def treatment(self):
        self.display_info()
        print(f"Disease  : {self.disease}")
        print(f"{self.name} is undergoing treatment.")


doc = Doctor("Rajan", 45, "Cardiology")
pat = Patient("Meena", 30, "Fever")

print("--- Doctor ---")
doc.diagnose()

print("\n--- Patient ---")
pat.treatment()
```

### Task 3 — Multilevel Inheritance: Employee → Developer → SeniorDeveloper

```python
class Employee:
    def __init__(self, name, employee_id, salary):
        self.name = name
        self.employee_id = employee_id
        self.salary = salary


class Developer(Employee):
    def __init__(self, name, employee_id, salary, programming_language):
        super().__init__(name, employee_id, salary)
        self.programming_language = programming_language

    def write_code(self):
        print(f"{self.name} is writing code in {self.programming_language}")


class SeniorDeveloper(Developer):
    def __init__(self, name, employee_id, salary, programming_language, experience):
        super().__init__(name, employee_id, salary, programming_language)
        self.experience = experience

    def review_code(self):
        print(f"{self.name} is reviewing code with {self.experience} years of experience")


sd = SeniorDeveloper("Karthik", "E201", 80000, "Python", 8)
print(f"Name       : {sd.name}")
print(f"Employee ID: {sd.employee_id}")
print(f"Salary     : {sd.salary}")
sd.write_code()
sd.review_code()
```

### Task 4 — Multiple Inheritance: Employee + Skills → Developer

```python
class Employee:
    def __init__(self, name, employee_id):
        self.name = name
        self.employee_id = employee_id

    def display_employee(self):
        print(f"Employee: {self.name} | ID: {self.employee_id}")


class Skills:
    def __init__(self, skill):
        self.skill = skill

    def display_skill(self):
        print(f"Skill: {self.skill}")


class Developer(Employee, Skills):
    def __init__(self, name, employee_id, skill, project):
        Employee.__init__(self, name, employee_id)
        Skills.__init__(self, skill)
        self.project = project

    def work(self):
        print(f"{self.name} is working on project '{self.project}'")


dev = Developer("Arun", "D101", "Python", "KGS Portal")
dev.display_employee()
dev.display_skill()
dev.work()
```

### Task 5 — Hierarchical Inheritance: Product → Electronics / Clothing

```python
class Product:
    def __init__(self, product_name, price):
        self.product_name = product_name
        self.price = price

    def display_product(self):
        print(f"Product: {self.product_name} | Price: Rs.{self.price}")


class Electronics(Product):
    def __init__(self, product_name, price, warranty):
        super().__init__(product_name, price)
        self.warranty = warranty

    def display_warranty(self):
        self.display_product()
        print(f"Warranty: {self.warranty} years")


class Clothing(Product):
    def __init__(self, product_name, price, size):
        super().__init__(product_name, price)
        self.size = size

    def display_size(self):
        self.display_product()
        print(f"Size: {self.size}")


e = Electronics("Laptop", 55000, 2)
c = Clothing("T-Shirt", 499, "L")

print("--- Electronics ---")
e.display_warranty()

print("\n--- Clothing ---")
c.display_size()
```

### Task 6 — Method Overriding: Employee Roles

```python
class Employee:
    def work(self):
        print("Employee is working")


class Developer(Employee):
    def work(self):
        print("Developer -> Writing code")


class Tester(Employee):
    def work(self):
        print("Tester -> Testing application")


class Manager(Employee):
    def work(self):
        print("Manager -> Managing team")


dev = Developer()
tester = Tester()
mgr = Manager()

dev.work()
tester.work()
mgr.work()
```

### Task 7 — KGS Software Company: Complete Employee Management System

```python
class Employee:
    def __init__(self, name, employee_id, salary):
        self.name = name
        self.employee_id = employee_id
        self.salary = salary

    def display_details(self):
        print(f"Name: {self.name} | ID: {self.employee_id} | Salary: Rs.{self.salary}")

    def work(self):
        print(f"{self.name} is working")


class Developer(Employee):
    def __init__(self, name, employee_id, salary, programming_language, project):
        super().__init__(name, employee_id, salary)
        self.programming_language = programming_language
        self.project = project

    def work(self):
        print(f"{self.name} is developing the {self.project} using {self.programming_language}")


class Tester(Employee):
    def __init__(self, name, employee_id, salary, testing_tool, project):
        super().__init__(name, employee_id, salary)
        self.testing_tool = testing_tool
        self.project = project

    def work(self):
        print(f"{self.name} is testing the {self.project} using {self.testing_tool}")


class HR(Employee):
    def __init__(self, name, employee_id, salary, department):
        super().__init__(name, employee_id, salary)
        self.department = department

    def work(self):
        print(f"{self.name} is managing the {self.department} department")


dev = Developer("Arun", "E001", 70000, "Python", "Employee Portal")
tester = Tester("Priya", "E002", 50000, "Selenium", "Employee Portal")
hr = HR("Meena", "E003", 45000, "HR")

dev.display_details()
dev.work()

print()
tester.display_details()
tester.work()

print()
hr.display_details()
hr.work()
```

## Abstraction
### Task 1: Food Ordering with Abstract Class

```python
from abc import ABC, abstractmethod


class Food(ABC):
    @abstractmethod
    def prepare(self):
        pass

    @abstractmethod
    def serve(self):
        pass


class Pizza(Food):
    def prepare(self):
        print("Pizza -> Preparing pizza")

    def serve(self):
        print("Pizza -> Serving pizza on a round plate")


class Burger(Food):
    def prepare(self):
        print("Burger -> Preparing burger")

    def serve(self):
        print("Burger -> Serving burger in a wrapper")


class Biryani(Food):
    def prepare(self):
        print("Biryani -> Preparing biryani")

    def serve(self):
        print("Biryani -> Serving biryani in a bowl")


for food in [Pizza(), Burger(), Biryani()]:
    food.prepare()
    food.serve()
    print()
```

### Task 2: Notification System

```python
from abc import ABC, abstractmethod


class Notification(ABC):
    @abstractmethod
    def send(self, message):
        pass


class EmailNotification(Notification):
    def send(self, message):
        print(f'Email -> Sending email: "{message}"')


class SMSNotification(Notification):
    def send(self, message):
        print(f'SMS -> Sending SMS: "{message}"')


class WhatsAppNotification(Notification):
    def send(self, message):
        print(f'WhatsApp -> Sending WhatsApp message: "{message}"')


msg = "Your order is confirmed"

for notifier in [EmailNotification(), SMSNotification(), WhatsAppNotification()]:
    notifier.send(msg)
```

## List Comprehension Tasks
### Task 1: Employee Salary Increment

```python
salaries = [18000, 22000, 25000, 30000, 35000]

incremented = [s + 3000 for s in salaries]
print(incremented)  # [21000, 25000, 28000, 33000, 38000]
```

### Task 2: Even Employee IDs

```python
employee_ids = [101, 102, 103, 104, 105, 106, 107, 108]

even_ids = [eid for eid in employee_ids if eid % 2 == 0]
print(even_ids)  # [102, 104, 106, 108]
```

### Task 3: Student Pass/Fail Status

```python
marks = [85, 42, 76, 35, 91, 28, 67]

status = ["Pass" if m >= 40 else "Fail" for m in marks]
print(status)  # ['Pass', 'Pass', 'Pass', 'Fail', 'Pass', 'Fail', 'Pass']
```

### Task 4: Product Price Discount

```python
prices = [1000, 2500, 5000, 7500, 10000]

discounted = [p * 0.9 for p in prices]
print(discounted)  # [900.0, 2250.0, 4500.0, 6750.0, 9000.0]
```
